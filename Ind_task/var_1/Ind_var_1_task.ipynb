## Велика кухня даних: що впливає на виручку ресторану?

У період високої конкуренції, зростання витрат на рекламу та змін у споживчій поведінці керівництво прагне виявити найбільш ефективні стратегії ведення ресторанного бізнесу. У розпорядженні аналітиків є набір даних, зібраний на основі інформації про діяльність 1000 ресторанів:  
- Number_of_Customers — кількість клієнтів, які відвідали ресторан за місяць.

- Monthly_Revenue — загальна виручка ресторану за місяць (у тис. доларів) — **цільова змінна для прогнозування.**
