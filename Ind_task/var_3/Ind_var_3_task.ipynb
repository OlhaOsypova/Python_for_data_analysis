### Аналіз поведінкових типів клієнтів та їх сегментація на основі даних  
Аналіз поведінкових типів клієнтів передбачає ґрунтовне дослідження профілів клієнтів компанії з метою глибшого розуміння їхньої поведінки, потреб та особливостей споживання. Такий підхід дозволяє бізнесу адаптувати свої продукти й маркетингові стратегії до різних груп клієнтів, враховуючи їхні відмінності у витратах, каналах покупок, реакції на акції та рівні залученості.

Замість просування продуктів на всю клієнтську базу, компанія може визначити сегменти, які з найбільшою ймовірністю зацікавляться конкретною пропозицією. Це дозволяє спрямувати маркетингові ресурси саме на цільові групи, підвищити ефективність кампаній та збільшити ймовірність успішного прийняття продукту.

Опис змінних у наборі даних

- ID - унікальний ідентифікатор клієнта.
- Year_Birth - рік народження клієнта.
- Education - найвищий рівень освіти клієнта.
- Marital_Status - сімейний стан клієнта.
- Income - річний дохід клієнта.
- Kidhome - кількість малих дітей у домогосподарстві.
- Teenhome - кількість підлітків у домогосподарстві.
- Recency - кількість днів з моменту останньої покупки.
- MntWines - витрати на вино.
- MntFruits - витрати на фрукти.
- MntMeatProducts - витрати на м’ясні продукти.
- MntFishProducts - витрати на рибні продукти.
- MntSweetProducts - витрати на солодощі.
- MntGoldProds - витрати на “золоті” товари.
- NumDealsPurchases - кількість покупок зі знижкою/в межах акцій.
- NumWebPurchases - кількість покупок через сайт.
- NumCatalogPurchases - кількість покупок через каталоги.
- NumStorePurchases - кількість покупок у фізичних магазинах.
- NumWebVisitsMonth - кількість відвідувань сайту за місяць.
- AcceptedCmp1 - AcceptedCmp5 - чи приймав клієнт відповідні маркетингові кампанії (0 - ні, 1 - так).
- Complain - чи подавав клієнт скаргу (0 - ні, 1 - так).
